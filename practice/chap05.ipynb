```python
from jupyturtle import forward, left, right, back

def draw(length):
    angle = 50
    factor = 0.6
    
    if length > 5:
        forward(length)
        left(angle)
        draw(factor * length)
        right(2 * angle)
        draw(factor * length)
        left(angle)
        back(length)
```

```python
draw(20)
```

```python
def kochcurve(length):
    if length < 3:
        # 基础情况：长度足够小时，才真正画直线
        forward(length)
    else:
        # 递归情况：每一段都不直接画线，而是调用自己
        kochcurve(length / 3)  # 第一段
        left(60)
        kochcurve(length / 3)  # 第二段
        right(120)
        kochcurve(length / 3)  # 第三段
        left(60)
        kochcurve(length / 3)  # 第四段
```

```python
from turtle import *

# 1. 定义递归函数
def sierpinski(length, order):
    if order == 0:
        # 基础情况：画一个等边三角形
        begin_fill()
        for _ in range(3):
            forward(length)
            left(120)
        end_fill()
    else:
        # 递归情况：每一阶都拆分为 3 个更小的谢尔宾斯基三角形
        # 1. 画左下角的小三角形
        sierpinski(length / 2, order - 1)
        
        # 2. 移动到顶点位置
        forward(length / 2)
        sierpinski(length / 2, order - 1)
        
        # 3. 移动到右下角位置
        backward(length / 2)
        right(60)
        forward(length / 2)
        left(60)
        sierpinski(length / 2, order - 1)
        
        # 4. 回到起始点（保持函数执行前后画笔位置一致）
        right(60)
        backward(length / 2)
        left(60)

# 2. 设置环境
speed(0) # 最快速度
penup()
goto(-200, -150) # 调整初始位置
pendown()
color("blue", "cyan") # 蓝边青色填充

# 3. 开始绘制（尝试 3 阶或 4 阶）
sierpinski(400, 3)

done()
```
