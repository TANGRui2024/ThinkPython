```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

```python
def shift_word(word,n):
    """Shift the letters of 'word' by 'n' places.
    >>> shift_word('cheer', 7)
    'jolly'
    >>> shift_word('melon', 16)
    'cubed'
    """
    letters = 'abcdefghijklmnopqrstuvwxyz'
    numbers = range(len(letters))
    letter_map = dict(zip(letters, numbers))
    res = ""
    for letter in word:
        new_letter = letters[(letter_map[letter] + n)%26]
        res += new_letter
    return res
```

```python
shift_word('cheer', 7)
```

```python
shift_word('melon',16)
```

```python
def value_count(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

```python
def most_frequent_letters(string):
    counter = value_count(string)
    items = counter.items()
    answer = sorted(items, key = lambda x: x[1], reverse = True)
    return answer
```

```python
word_list = open('words.txt').read().split()
```

```python
def sort_word(word):
    return ''.join(sorted(word))
```

```python
def find_anagrams():
    anagram_map={}
    for word in word_list:
        signature = sort_word(word)
        
        if signature not in anagram_map:
            anagram_map[signature] = []
            anagram_map[signature].append(word)
    return anagram_map

all_anagrams = find_anagrams()
```

```python
find_anagrams()
```

```python
def value_length(pair):
    key, value = pair
    return len(value)
```

```python
anagram_items = sorted(all_anagrams.items(), key = value_length)
for key, value in anagram_items[-10:]:
    print(value)
```

```python
longest = 7

for key, value in anagram_items:
    if len(value) > 1:
        word_len = len(value[0])
        if word_len > longest:
            longest = word_len
            print(value)
```

```python
def word_distance(word1, word2):
    """Computes the number of places where two word differ.
    >>> word_distance("hello", "hxllo")
    1
    >>> word_distance("ample", "apply")
    2
    >>> word_distance("kitten","mutton")
    3
    """
    differ = 0
    zip(word1, word2)
    for code1, code2 in zip(word1,word2):
        if code1 != code2:
            differ += 1
    return differ
        
        
```
