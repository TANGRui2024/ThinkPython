```python
def is_anagram(word1, word2):
    """Checks whether two words are anagrams.
    
    >>> is_anagram('tops', 'stop')
    True
    >>> is_anagram('skate','takes')
    True
    >>> is_anagram('tops','takes')
    False
    >>> is_anagram('skate','stop')
    False
    """
    return sorted(word1) == sorted(word2)
```

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
    
run_doctests(is_anagram)
```

```python
def reverse_word(word):
    return ''.join(reversed(word))
```

```python
def is_palindrome(word):
    """Check if a word is a palindrome.
    >>> is_palindrome('bob')
    True
    >>> is_palindrome('alice')
    False
    >>> is_palindrome('a')
    True
    >>> is_palindrome('')
    True
    """
    
    if word == reverse_word(word):
        return True
    else:
        return False
```

```python
 from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
    
run_doctests(is_palindrome)
```

```python
for word in word_list:
    if len(word) >= 7 and is_palindrome(word):
        print(word)
```

```python
def reverse_sentence(input_string):
    
    words = input_string.split()
    reversed_list = list(reversed(words))
    
    new_sentence = ' '.join(reversed_list)
    
    return new_sentence.capitalize()
```
