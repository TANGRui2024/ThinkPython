```python
def md5_digest(filename):
    data = open(filename, 'rb').read()
    md5_hash = hashlib.md5()
    md5_hash.update(data)
    digest = md5_hash.hexdigest()
    return digest
```

```python
def replace_all(old, new, source_path, dest_path):
    
    reader = open(source_path)
    content = reader.read()
    reader.close()
    content = content.replace(old,new)
    writer = open(dest_path, 'w')
    writer.write(content)
    writer.close()

    
```

```python
# Solution goes here

import os

def is_image(path, extensions):
    """
    判断路径 path 是否以 extensions 列表中的某个后缀结尾
    """
    base, ext = os.path.splitext(path)
    
    if ext.lower() in extensions:
        return True
    else:
        return False
```

```python
# Solution goes here

def add_path(path, sf):
    
    reader = open(path,'rb')
    content = reader.read()
    reader.close()
    md5_hash = hashlib.md5()
    md5_hash.update(content)
    digest = md5_hash.hexdigest()
    
    if digest in sf:
        temp_list = sf[digest]
        temp_list.append(path)
        sf[digest]  = temp_list
    else:
        sf[digest] = [path]
    
```

```python
# Solution goes here

def walk_images(dirname,sf):
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)
        if is_image(path, extensions):
            add_path(path,sf)
        elif os.path.isdir(path):
            walk_images(path, sf)
        
```
