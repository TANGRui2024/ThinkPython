```python
from diagram import Frame, Stack, make_binding

main = Frame([], name='__main__', loc='left')
frames = [main]

ns = 3, 2, 1
recurses = 2, 1, 1
results = 6, 2, 1

for n, recurse, result in zip(ns, recurses, results):
    binding1 = make_binding('n', n)
    binding2 = make_binding('recurse', recurse)
    frame = Frame([binding1, binding2], 
                  name='factorial', value=result,
                  loc='left', dx=1.2)
    frames.append(frame)
    
binding1 = make_binding('n', 0)
frame = Frame([binding1], name='factorial', value=1, 
              shim=1.2, loc='left', dx=1.4)
frames.append(frame)

stack = Stack(frames, dy=-0.45)
```

```python
def factorial(n):
    space = ' ' * (4*n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
```

```python
def factorial_v2(n):
    # 用缩进来表示当前的“嵌套深度”
    space = '    ' * n
    print(f"{space}🚀 进入第 n={n} 层")

    if n == 0:
        print(f"{space}✅ 到达底层(n=0)，准备返回接力棒: 1")
        return 1
    else:
        # 核心：在这里停下来，去问下一层
        print(f"{space}🔍 n={n} 层正在等待 n={n-1} 层的结果...")
        recurse = factorial_v2(n-1)
        
        # 重点：这一行就是你刚才困惑的地方，它是“收到”结果后的动作
        result = n * recurse
        print(f"{space}📥 n={n} 层收到了接力棒({recurse})，计算 {n} * {recurse} = {result}")
        print(f"{space}📤 n={n} 层完成任务，向上返回: {result}")
        return result

# 让我们运行一下
print("--- 开始追踪 factorial(3) ---")
final_result = factorial_v2(3)
print(f"--- 最终结果是: {final_result} ---")
```

```python
import math # 导入数学库以使用 sqrt 函数

def calculate_distance(x1, y1, x2, y2):
    # 1. 计算两个点在横轴(x)上的差值
    dx = x2 - x1
    
    # 2. 计算两个点在纵轴(y)上的差值
    dy = y2 - y1
    
    # 3. 根据勾股定理计算距离
    # dx**2 表示 dx 的平方
    distance = math.sqrt(dx**2 + dy**2)
    
    return distance

# --- 测试一下 ---
# 比如点 (0, 0) 和点 (3, 4) 之间的距离应该是 5
result = calculate_distance(0, 0, 3, 4)
print(f"两点间的距离是: {result}")
```

```python
def ackermann(x,y):
    if x == 0:
        return y +1
    elif x > 0 and y == 0:
        return ackermann(x-1,1)
    elif x > 0 and y >0:
        return ackermann(x-1,ackermann(x,y-1))
```

```python
def gcd(a, b):
    # 基准情况：如果除数 b 变成了 0
    if b == 0:
        # 根据公式 gcd(a, 0) = a，返回 a
        return a
    
    # 计算余数 r
    r = a % b
    
    # 递归步骤：返回 gcd(b, r) 的结果
    # 记得要写 return 才能把深层计算的结果传回来
    return gcd(b, r)
```
