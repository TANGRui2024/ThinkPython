```python
def repeat(word,n):
    print(word*n)
```

```python
spam = 'Spam,'
repeat(spam,4)
```

```python
def first_two_lines():
    repeat(spam,4)
    repeat(spam,4)
```

```python
first_two_lines()
```

```python
def last_three_lines():
    repeat(spam,2)
    print('(Lovely Spam, Wonderful Spam!)')
    repeat(spam,2)
```

```python
last_three_lines()
```

```python
def print_verse():
    first_two_lines()
    last_three_lines()
```

```python
print_verse()
```

```python
def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
```

```python
line1 = "Hello haha"
line2 = " world!"
cat_twice(line1, line2)
```

```python
from diagram import make_frame, Stack

d1 = dict(line1=line1, line2=line2)
frame1 = make_frame(d1, name='__main__',dy = -0.3, loc='left')

d2 = dict(part1=line1, part2=line2, cat = line1+line2)
frame2 = make_frame(d2, name = 'cat_twice', dy = -0.3, loc = 'left')

d3 = dict(string= line1+line2)
frame3 = make_frame(d3, name = 'print_twice',
                    offsetx = 0.04, offsety=-0.3, loc = 'left')

d4 = {"?":line1 + line2}
frame4 = make_frame(d4, name = 'print',
                    offsetx = 0.5, offsety = 0, loc = 'left')

stack = Stack([frame1, frame2, frame3, frame4], dy = -0.8)
```

```python
from diagram import diagram, adjust

width, height, x, y = [5, 2, 1, 1]
ax = diagram(width,height)
bbox = stack.draw(ax,x,y)

import matplotlib.pyplot as plt
plt.savefig('chap03_stack_diagram.png',dpi=300)
```

```python
text = 'W'
height = 5

def triangle(text, height):
    for i in range(height+1):
        spaces = ' ' * (height - i)
        content = text * (2*i +1)   
        print(spaces+content)
        
```

```python
triangle(text,20)
```
